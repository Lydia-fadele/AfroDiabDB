```python

from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd

def calculate_descriptors(smiles):
    try:
        mol = Chem.MolFromSmiles(str(smiles))

        if mol is None:
            return pd.Series({
                "Molecular_Weight": None,
                "LogP": None,
                "HBA": None,
                "HBD": None,
                "TPSA": None,
                "Rotatable_Bonds": None
            })

        return pd.Series({
            "Molecular_Weight": round(Descriptors.MolWt(mol), 2),
            "LogP": round(Descriptors.MolLogP(mol), 2),
            "HBA": Descriptors.NumHAcceptors(mol),
            "HBD": Descriptors.NumHDonors(mol),
            "TPSA": round(Descriptors.TPSA(mol), 2),
            "Rotatable_Bonds": Descriptors.NumRotatableBonds(mol)
        })

    except:
        return pd.Series({
            "Molecular_Weight": None,
            "LogP": None,
            "HBA": None,
            "HBD": None,
            "TPSA": None,
            "Rotatable_Bonds": None
        })

descriptor_df = df["Canonical_SMILES"].apply(calculate_descriptors)

df.update(descriptor_df)

print("Descriptor calculation completed.")
```

# Phase 1C: Lipinski Rule of Five Evaluation

Lipinski's Rule of Five was applied to assess the drug-likeness of each phytochemical in AfroDiabDB.

The following criteria were evaluated:

- Molecular Weight ≤ 500 Da
- LogP ≤ 5
- Hydrogen Bond Acceptors (HBA) ≤ 10
- Hydrogen Bond Donors (HBD) ≤ 5

Compounds with more than one violation were classified as non-compliant.

```python

def lipinski_violations(row):
    violations = 0

    if row["Molecular_Weight"] > 500:
        violations += 1

    if row["LogP"] > 5:
        violations += 1

    if row["HBA"] > 10:
        violations += 1

    if row["HBD"] > 5:
        violations += 1

    return violations

df["Lipinski_Violations"] = df.apply(lipinski_violations, axis=1)

df["Lipinski_Status"] = df["Lipinski_Violations"].apply(
    lambda x: "Pass" if x <= 1 else "Fail"
)

print("Lipinski evaluation completed successfully.")
```

# Phase 2D: Drug-Likeness Assessment

Lipinski's Rule of Five is widely used as a preliminary filter for evaluating the oral drug-likeness of small molecules. The rule considers molecular weight, lipophilicity (LogP), hydrogen bond acceptors (HBA), and hydrogen bond donors (HBD) to identify compounds with favorable physicochemical properties for oral bioavailability.

The distribution of Lipinski-compliant and non-compliant phytochemicals in AfroDiabDB was determined to assess the overall drug-likeness profile of the database.

```python
df["Lipinski_Status"].value_counts()
```

## Interpretation

Lipinski's Rule of Five analysis revealed that 43 of the 53 phytochemicals (81.1%) satisfied the drug-likeness criteria, while 10 compounds (18.9%) exhibited more than one Lipinski violation and were classified as non-compliant. The high proportion of compliant compounds indicates that AfroDiabDB contains a substantial number of phytochemicals with physicochemical properties consistent with oral drug-like molecules.

These findings support the utility of AfroDiabDB as a valuable resource for virtual screening, molecular docking, pharmacokinetic prediction, and natural product-based drug discovery targeting diabetes mellitus.
